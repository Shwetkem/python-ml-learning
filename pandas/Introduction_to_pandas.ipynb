## Pandas Datatypes: Series & DataFrame

`Series` = 1-dimensional data (like a single column). `DataFrame` = 2-dimensional data (rows + columns), built by combining Series together.

```python
# 2 main datatypes
series = pd.Series(["BMW","Mercedes","Toyota"])
```

## Handling Missing Values

Two common strategies: filling missing values with `.fillna()` (e.g. using the column mean), or removing rows with missing values entirely using `.dropna()`.

```python
car_sales_missing_values = pd.read_csv("car_sales_missing_values.csv")
```

```python
car_sales_missing_values
```

```python
car_sales_missing_values["Odometer (KM)"].fillna(car_sales_missing_values["Odometer (KM)"].mean(numeric_only = True),inplace = True)
```

```python
car_sales_missing_values["Odometer (KM)"].mean(numeric_only = True)
```

## Creating, Modifying and Removing Columns

Different ways to add a new column — from a Series, a plain Python list, a calculation on existing columns, or a single constant value — and how to drop a column with `.drop()`.

```python
# Column creation from Series
seats_column = pd.Series([5,5,5,5,5])

# New column called seats
car_sales["Seats"] = seats_column
car_sales
```

```python
car_sales["Seats"] = car_sales["Seats"].fillna(car_sales["Seats"].mean())
```

```python
 car_sales
```

```python
# Column from Python list
fuel_economy = [7.5, 9.2, 5.0, 9.6, 8.7, 6.8, 9.0, 4.3, 8.6, 5.5]  #If creating a column from a python list, the list must be the same size as that of the rows in dataframe, the same is not applicable in series
car_sales["Mileage per 100KM"] = fuel_economy
car_sales
```

```python
# Creating column from other columns
car_sales["Total fuel used"] = (car_sales["Odometer"]/100)*car_sales["Mileage"]
car_sales
```

```python
# Create a column from a single value
car_sales["Number of wheels"] = 4
car_sales
```

```python
car_sales["Passed safety Test"] = True   # Creates a column of a bool type
car_sales
```

```python
car_sales.dtypes
```

```python
car_sales = car_sales.drop("Total fuel used", axis = 1) # Removing columns from the dataframe
car_sales
```

## Shuffling and Sampling Data

Using `.sample(frac=...)` to shuffle or take a random subset of rows, `.reset_index()` to clean up the resulting index, and `.apply()` with a `lambda` to transform column values.

```python
# Shuffling the data
car_sales_shuffled = car_sales.sample(frac=1)
car_sales_shuffled
```

```python
car_salezzzz = pd.read_csv("car_sales.csv")
```

```python
car_salezzzz
```

```python
# Only shuffling 20% of data
car_sales_shuffled = car_salezzzz.sample(frac = 0.2)
car_sales_shuffled
```

```python
car_sales_shuffled.reset_index(drop=True, inplace=True)
```

```python
car_sales_shuffled = pd.read_csv("car_sales.csv")
```

```python
car_sales_shuffled
```

```python
#Applying lambda functions to change the column values
car_sales_shuffled["Odometer"] = car_sales_shuffled["Odometer"].apply(lambda x: x/1.6)
car_sales_shuffled
```
