## Matplotlib example workflow

```python
# 0. import matplotlib and get it ready for plotting in Jupyter
%matplotlib inline
import matplotlib.pyplot as plt

# 1. Prepare data
x = [1, 2, 3, 4]
y = [90, 20, 30, 40]

# 2. Setup plot
fig, ax = plt.subplots(figsize=(7,5))   # figsize here describes the (width, height) of the plot exactly in that order

# 3. Plot data
ax.plot(x, y);  # add data

# 4. Customise plot
ax.set(title = "Simple plot",
       xlabel = "x-axis",
       ylabel = "y-axis"); 

# 5. Save & show (we save the whole figure)
fig.savefig(r"C:\Users\lenovo\Desktop\sample_project_1\images")
```

# Two options for subplots

```python
# Subplot option 1
y = np.array([45,67,32,89,200])
fig, ((ax1, ax2),(ax3, ax4)) = plt.subplots(nrows=2,ncols=2,figsize=(10,5))

# Plot to each different axis
ax2.plot(y);
ax1.hist(np.random.randn(100));
ax3.scatter(y, np.sin(y));
ax4.bar(nut_butter_prices.keys(), nut_butter_prices.values());
```

```python
# Subplots option 2
fig, ax = plt.subplots(nrows = 2, ncols = 2, figsize = (10,5))

# Plot to each different index
ax[0, 0].plot(x, x/2);
ax[0, 1].bar(nut_butter_prices.keys(), nut_butter_prices.values())
ax[1, 0].hist(np.random.randn(1000));
ax[1, 1].scatter(x, np.sin(x))
```

### Which one must one use? (pyplot vs matplotlib OO (object-oriented) method)
* When plotting something quickly, okay to use the pyplot method
* When plotting something more advanced, use the OO method

```python
over_50 = heart_disease
over_50.head()
```

```python
over_50 = heart_disease[heart_disease["age"] > 50]
len(over_50)
# len(heart_disease)
```

```python
over_50.head()
```

```python
# Pyplot method
over_50.plot.scatter("age", "chol", c="target");
over_50.plot("age","chol", kind="scatter",c="target");
```

```python
fig, ax = plt.subplots(1,2)
ax[0].scatter(over_50["age"], over_50["chol"]);
ax[1].scatter(over_50["age"], over_50["thalach"]);
```

```python
# OO method mixed with pyplot method
fig, ax = plt.subplots()
over_50.plot(kind="scatter", x="age", c = "target", y="chol", ax = ax);

ax.set_xlim([45, 100])  # sets the limits
```

```python
over_50.target.values
```

```python
# OO method from scratch
fig, ax = plt.subplots(figsize = (10, 6))

# PLotting the data
scatter_plot = ax.scatter(x = over_50["age"],
                     y = over_50["chol"],
                     c = over_50["target"]);

# Customize the plot
ax.set(title = "Heart Disease and Cholestrol Levels",
       xlabel = "Age",
       ylabel = "Choleterol");

# Add a legend 
ax.legend(*scatter_plot.legend_elements(), title = "Target");

# Add a horizontal line
ax.axhline(over_50["chol"].mean(), linestyle = "--");
```

```python
# Set the style
plt.style.use("seaborn-v0_8-whitegrid")

# OO method from scratch
fig, ax = plt.subplots(figsize = (10, 6))

# PLotting the data
scatter_plot = ax.scatter(x = over_50["age"],
                     y = over_50["chol"],
                     c = over_50["target"],
                     cmap = "winter");

# Customize the plot
ax.set(title = "Heart Disease and Cholestrol Levels",
       xlabel = "Age",
       ylabel = "Choleterol");

# Add a legend 
ax.legend(*scatter_plot.legend_elements(), title = "Target");

# Add a horizontal line
ax.axhline(over_50["chol"].mean(), linestyle = "--");
```

This plot shows some information about the heart disease dataset....
    

```python
# Customizing the y and x axis limitations
fig, ax = plt.subplots(2,1,figsize=(10,8), sharex = True)
scatter0 = ax[0].scatter(x = over_50["age"], y = over_50["chol"], c = over_50["target"], cmap="winter")
scatter1 = ax[1].scatter(x = over_50["age"], y = over_50["thalach"], c = over_50["target"], cmap="winter")

ax[0].axhline(y = over_50["chol"].mean(), linestyle = "--", color="red");
ax[1].axhline(y = over_50["thalach"].mean(), linestyle = "--",color="red");

ax[0].set(title = "Heart Disease and Cholesterol Levels", ylabel = "Cholesterol");
ax[0].set_xlim([50, 80]) # changing the x-axis limits and as both plots share the same x-axis it works on both of them
ax[0].set_ylim([100, 420]) 

ax[1].set_ylim([90,220])
ax[1].set(title = "Heart Disease and Max Heart Rate", xlabel = "Age", ylabel = "Thalach");   # sharex is true yet its not working figure it out later

ax[0].legend(*scatter0.legend_elements(), title="Target");
ax[1].legend(*scatter1.legend_elements(), title="Target");

# Add a title to the figure
fig.suptitle("Heart Disease Analysis", fontsize = 16, fontweight = "bold");
```
